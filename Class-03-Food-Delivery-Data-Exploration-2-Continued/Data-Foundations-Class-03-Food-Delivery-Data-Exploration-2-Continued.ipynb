## Arange and steps

`01_arange_and_steps.py`

np.arange(start, end, step) makes numbers in a row. The end is left out.
A slice can take a step too: array[start : end : step]. A negative step walks backwards.


```python
import numpy as np

# The app team: IDs for the first 1000 customers in a new city
ids = np.arange(1, 1001)
print(ids[:5])
print(ids[-3:])
print(ids.size)

print(np.arange(1, 1000)[-1])      # 999: the end is left out

# The third number is the step
print(np.arange(100, 1000, 100))   # the product team: cost-for-two filter buttons, 100 to 900
print(np.arange(3, 5, 0.5))        # rating filter 3.0 to 4.5: a decimal step works; Python's range() does not allow it

# Step slicing
votes = np.array([ 775,  787,  918,   88,  166,  286, 2556,  324,  504,  402])
a = np.arange(10)    # positions of our 10 restaurants
print(a)
print(a[::2])        # every second element: the quality team visits these
print(votes[::2])    # the votes of those five restaurants
print(a[::-1])       # reversed: newest first, as on the "Your orders" screen
print(a[::-2])       # from the end, every second element
print(a[::-3])       # [9 6 3 0]
print(a[1:8:3])      # [1 4 7]: start at 1, jump by 3, stop before 8
```

## Filtering with masks

`03_filtering_with_masks.py`

A mask is an array of True and False. Inside the brackets it filters:
only the elements where the mask is True come through.


```python
import numpy as np

votes = np.array([ 775,  787,  918,   88,  166,  286, 2556,  324,  504,  402])
costs = np.array(["'800.0'" ,"'800.0'", "'800.0'", "'300.0'", "'600.0'", "'600.0'", "'600.0'", "'700.0'" ,"'550.0'", "'500.0'"])
clean_costs = np.char.strip(costs, "'").astype(float)

# The product team: a "Popular" filter, 500 votes or more
print(votes >= 500)              # the mask: one True or False for each restaurant
print(votes[votes >= 500])       # the filter: only the votes that pass
print((votes >= 500).sum())      # True counts as 1, so this counts how many passed: 5

# The manager: are popular places more expensive?
# A mask from one array can filter another, when the two arrays line up
print(clean_costs[votes >= 500])
print(clean_costs[votes >= 500].mean())   # popular places: 710.0
print(clean_costs[votes < 500].mean())    # the rest: 540.0

# Marketing: a "Popular and affordable" collection
# Two conditions: each in round brackets, joined by &
mask = (votes > 500) & (clean_costs < 700)
print(votes[mask])
print(clean_costs[mask])

# Your turn (another business, the same tool): a fitness band records the mood and the step count of each user
mood = np.array(["Happy", "Sad", "Happy", "Sad",
                 "Happy", "Happy", "Sad", "Happy"])
steps = np.array([9200, 4100, 10500, 5200,
                  8700, 11000, 3900, 9600])
print(steps[mood == "Happy"].mean())
print(steps[mood == "Sad"].mean())
```

## Column stack and reshape

`04_column_stack_and_reshape.py`

np.column_stack puts arrays side by side as columns: one row per restaurant.
reshape refills the same numbers into new rows and columns. rows x columns must equal the size.
-1 lets NumPy work out one side.


```python
import numpy as np

votes = np.array([ 775,  787,  918,   88,  166,  286, 2556,  324,  504,  402])
costs = np.array(["'800.0'" ,"'800.0'", "'800.0'", "'300.0'", "'600.0'", "'600.0'", "'600.0'", "'700.0'" ,"'550.0'", "'500.0'"])
clean_costs = np.char.strip(costs, "'").astype(float)

# The manager: one sheet, one row per restaurant
data = np.column_stack((votes, clean_costs))
print(data.shape)       # (10, 2): 10 restaurants, 2 columns
print(data[:3])         # the votes became floats: one type only

# The app team: restaurant cards in a grid on the home screen
ids = np.arange(1, 10)              # 9 restaurant IDs
print(ids.reshape(3, 3))            # 3 cards in each row, in reading order

ids = np.arange(1, 11)              # 10 restaurant IDs
print(ids.reshape(-1, 2))           # a phone: 2 cards in a row, NumPy works out 5 rows
print(ids.reshape(-1, 5))           # a tablet: 5 cards in a row, NumPy works out 2 rows
# ids.reshape(3, 3)    ValueError: cannot reshape array of size 10 into shape (3,3)
# ids.reshape(-1, 3)   ValueError: cannot reshape array of size 10 into shape (3)

# A model in scikit-learn wants a table: one row per sample, one column per feature.
X = votes.reshape(-1, 1)
print(X.shape)          # (10, 1)

# .T and reshape give the same shape, not the same table
a = np.arange(1, 7).reshape(2, 3)
print(a.T)
print(a.reshape(3, 2))

# On the manager's sheet: .T keeps the votes together, reshape mixes votes and rupees
print(data.T[0, :4])
print(data.reshape(2, 10)[0, :4])
```

## Aggregates and axis

`05_aggregates_and_axis.py`

Aggregates turn many numbers into one: sum, mean, max, min, std.
On a table, axis=0 gives one answer per column (the rows disappear),
axis=1 gives one answer per row (the columns disappear).


```python
import numpy as np

votes = np.array([ 775,  787,  918,   88,  166,  286, 2556,  324,  504,  402])
costs = np.array(["'800.0'" ,"'800.0'", "'800.0'", "'300.0'", "'600.0'", "'600.0'", "'600.0'", "'700.0'" ,"'550.0'", "'500.0'"])
clean_costs = np.char.strip(costs, "'").astype(float)
data = np.column_stack((votes, clean_costs))

# The manager's weekly slide
print(data[:, 0].sum())     # total votes
print(data[:, 1].mean())    # average cost for two
print(data[:, 0].max())     # most votes
print(data[:, 1].min())     # cheapest

# Operations: can a restaurant promise "30 minutes"? Same average, different spread
a = np.array([28, 30, 32])      # restaurant A, last three deliveries, in minutes
b = np.array([10, 30, 50])      # restaurant B
print(a.mean(), b.mean())
print(a.std().round(2), b.std().round(2))

# Operations: orders of 3 restaurants (rows) on Fri, Sat, Sun (columns)
orders = np.array([[20, 35, 30],
                   [12, 18, 15],
                   [25, 40, 32]])
print(orders.sum())               # 227: everything
print(np.sum(orders, axis=0))     # one total per day: plan riders for the busiest
print(np.sum(orders, axis=1))     # one total per restaurant

print(data.mean(axis=0))    # average votes and average cost, in one line
print(data.max(axis=0))     # the top of each column: not one restaurant

# Your turn (a restaurant owner): the average votes of the restaurants that cost 800 for two
print(votes[clean_costs == 800].mean())
```

## Any all where

`06_any_all_where.py`

np.any: is at least one True? np.all: is every one True? Each gives one True or False.
np.where(condition, a, b) is an if-else for every element.
np.where(condition) gives the positions where the condition is True.


```python
import numpy as np

votes = np.array([ 775,  787,  918,   88,  166,  286, 2556,  324,  504,  402])
costs = np.array(["'800.0'" ,"'800.0'", "'800.0'", "'300.0'", "'600.0'", "'600.0'", "'600.0'", "'700.0'" ,"'550.0'", "'500.0'"])
clean_costs = np.char.strip(costs, "'").astype(float)

# The manager: check the data before the weekly slide
print(np.any(clean_costs > 3000))
print(np.all(clean_costs < 5000))

ratings = np.array([4.5, 4.3, 4.1])
print(np.all(ratings > 4.2))        # a "Top rated" badge: 4.1 fails, so False

ratings = np.array([4.9, 4.1, 4.8])     # delivery partners' ratings
print(np.where(ratings >= 4.2, "Green Flag", "Red Flag"))

tags = np.where(votes >= 500, "Popular", "New")    # a badge for every card
print(tags)

print(np.where(votes >= 500))           # the positions
print(np.where(clean_costs > 1000))     # no match: an empty array
```

## Sorting

`07_sorting.py`

np.sort gives a sorted copy. Reverse it with [::-1] for the largest first.
np.argsort gives the positions in sorted order: use them on a parallel array.
In 2D, axis=1 sorts each row and axis=0 sorts each column, on its own.


```python
import numpy as np

votes = np.array([ 775,  787,  918,   88,  166,  286, 2556,  324,  504,  402])
costs = np.array(["'800.0'" ,"'800.0'", "'800.0'", "'300.0'", "'600.0'", "'600.0'", "'600.0'", "'700.0'" ,"'550.0'", "'500.0'"])
clean_costs = np.char.strip(costs, "'").astype(float)
data = np.column_stack((votes, clean_costs))

print(np.sort(votes))
print(np.sort(votes)[::-1])
print(np.argsort(votes))

# The manager: the most popular places, and what they cost
order = np.argsort(votes)[::-1]     # most votes first
print(votes[order][:3])
print(clean_costs[order][:3])       # the costs stay with their restaurants

m = np.array([[34, 11,  8],
              [ 7, 45, 18],
              [ 9, 23, 20]])
print(np.sort(m, axis=1))
print(np.sort(m, axis=0))

print(np.sort(data, axis=0)[:3])            # each column sorted apart: rows broken
print(data[np.argsort(data[:, 0])][:3])     # whole rows, ordered by votes

# Numbers stored as text sort like words: a "low to high" sort puts 1200 first
prices = np.array(["800", "1200", "300", "550"])
print(np.sort(prices))
```
